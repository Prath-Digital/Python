### Question 1
Install and import Seaborn.

```python
%pip install seaborn matplotlib pandas

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

### Question 2
Load Seaborn’s built-in dataset `tips` or `iris`.

```python
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

### Question 3
Display the first 5 rows and check dataset info.

```python
print(tips.head())
print(tips.info())
```

### Question 4
Create a histogram of a numerical column (e.g., total_bill from tips).

```python
sns.histplot(data=tips, x='total_bill', bins=20, kde=False)
plt.show()
```

### Question 5
Create a KDE plot of a numerical column.

```python
sns.kdeplot(data=tips, x='total_bill', fill=True)
plt.show()
```

### Question 6
Combine histogram and KDE into a single distplot or displot.

```python
sns.histplot(data=tips, x='total_bill', bins=20, kde=True)
plt.show()
```

### Question 7
Create a boxplot comparing total_bill by day.

```python
sns.boxplot(data=tips, x='day', y='total_bill')
plt.show()
```

### Question 8
Create a violinplot for tip amount by gender.

```python
sns.violinplot(data=tips, x='sex', y='tip')
plt.show()
```

### Question 9
Use stripplot or swarmplot to visualize the spread of tips by smoker status.

```python
sns.stripplot(data=tips, x='smoker', y='tip', jitter=True)
plt.show()
```

### Question 10
Plot a scatterplot between total_bill and tip.

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.show()
```

### Question 11
Add hue, size, and style to show smoker and gender information.

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='smoker', style='sex', size='size')
plt.show()
```

### Question 12
Use lineplot to show trend of tips by day.

```python
sns.lineplot(data=tips, x='day', y='tip', estimator='mean')
plt.show()
```

### Question 13
Create a heatmap for correlation between numeric columns in the dataset.

```python
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

### Question 14
Use annotations and color maps for better clarity.

```python
sns.heatmap(corr, annot=True, cmap='viridis', linewidths=0.5)
plt.show()
```

### Question 15
Create a clustermap to group similar rows/columns.

```python
%pip install scipy

sns.clustermap(corr, annot=True, cmap='coolwarm')
plt.show()
```

### Question 16
Use regplot() to show linear relationship between variables.

```python
sns.regplot(data=tips, x='total_bill', y='tip')
plt.show()
```

### Question 17
Use lmplot() for multiple regression lines with categories (hue).

```python
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
plt.show()
```

### Question 18
Set a different Seaborn theme (e.g., darkgrid, whitegrid).

```python
sns.set_theme(style='darkgrid')
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.show()
```

### Question 19
Customize color palette (e.g., palette='pastel').

```python
sns.set_palette('pastel')
sns.boxplot(data=tips, x='day', y='total_bill')
plt.show()
```

### Question 20
Apply context settings for publication-style plots.

```python
sns.set_context('paper')
sns.histplot(data=tips, x='total_bill', bins=20, kde=True)
plt.show()
```
